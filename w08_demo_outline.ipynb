## Demo Outline — 5 Minutes

**Question (30 sec)**
FlyRank clients have hundreds of pages losing clicks. Which ones should
a content team refresh first? Can GSC signals answer that?

**Method (90 sec)**
- Data: 79M rows, FlyRank internship warehouse, month=2026-03
- Label: did clicks grow ≥20% the next month?
- Model: Random Forest on 5 GSC features
- Validation: client-grouped split (no client in both train and test)

**One chart (60 sec)**
Model vs Baseline AUC:
- Hand-written rule: 0.64
- Random Forest:     0.82
Impressions volume and data coverage drove the gap — not CTR or position.

**One honest result (60 sec)**
The model surfaces a ranked queue of 75,869 pages.
False negatives are high (35k missed) — the model is precision-first,
not recall-first. It finds reliable candidates, not every opportunity.

**One recommendation (60 sec)**
Use the top 50 as a weekly human-review triage list.
Start with LOW_CTR_GOOD_POSITION pages — low editorial cost, high
potential click gain, 30–60 minutes per page.
